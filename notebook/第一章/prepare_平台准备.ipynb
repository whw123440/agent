```python
# 1. 测试天气 api的获取 
import requests
response=requests.get("https://wttr.in/Beijing?format=j1")
print(   "天气API响应状态码:",response.status_code)
print(   "天气API响应内容:",response.text)

```

```python
from tavily import TavilyClient
import os

key=os.environ.get("TAVILY_API_KEY")   # 可以将 API_KEY 放到 .env 文件中， 然后在 .vscode/settings.json 中配置环境变量文件加载
client = TavilyClient(api_key=key)

try:
    response = client.search("北京天气")
    print(response)
except Exception as e:
    print("Tavily API 调用失败:", e)
```

```python
# # 测试 LLM API -  我们先用: AIHubmix
from openai import OpenAI
import os

open_ai_key=os.environ.get("OPENAI_API_KEY")
model_name=os.environ.get("MODEL_NAME")
openai_base_url=os.environ.get("OPENAI_BASE_URL")

print( open_ai_key, model_name, openai_base_url )

# 创建一个OpenAI客户端实例
client = OpenAI(
    api_key=open_ai_key,
    base_url=openai_base_url
)
try:
    response=client.chat.completions.create(
        model=model_name,
        messages=[
            {"role": "user", "content": "你好,你是哪个模型?开发商是谁?"}
        ],
        max_tokens=1024,
        temperature=0.5
    )
    print(  "LLM API连接成功，响应:", response.choices[0].message.content )
except Exception as e:
    print(  "LLM API连接失败，错误信息:", e )
```

```python
# 2.测试Tavily API ，他是一个给智能体提供搜索能力的API，类似于一个搜索引擎。我们可以用它来获取一些信息，比如天气、新闻等。
from tavily import TavilyClient
import os

key=os.environ.get("TAVILY_API_KEY")   # 可以将 API_KEY 放到 .env 文件中， 然后在 .vscode/settings.json 中配置环境变量文件加载
client = TavilyClient(api_key=key)

try:
    response = client.search("长沙在Patchy rain nearby， 气温31摄氏度天气下的旅游景点推荐及理由", search_depth="advanced", include_answers=True)
    print(response)
except Exception as e:
    print("Tavily API 调用失败:", e)
```
